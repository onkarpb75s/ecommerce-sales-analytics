# E-Commerce Sales & Customer Analytics
Portfolio analysis using Python and Pandas. The dataset is synthetic and reproducible.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/orders.csv', parse_dates=['order_date'])
df.head()
```

```python
kpis = {
    'Revenue': df.revenue.sum(),
    'Profit': df.profit.sum(),
    'Orders': df.order_id.nunique(),
    'Customers': df.customer_id.nunique(),
    'AOV': df.revenue.sum() / df.order_id.nunique(),
    'Profit Margin': df.profit.sum() / df.revenue.sum(),
}
pd.Series(kpis)
```

```python
monthly = df.assign(month=df.order_date.dt.to_period('M').astype(str)).groupby('month').revenue.sum()
monthly.plot(figsize=(10,4), marker='o', title='Monthly Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
category = df.groupby('category').agg(revenue=('revenue','sum'), profit=('profit','sum')).sort_values('revenue', ascending=False)
category
```

```python
customer_orders = df.groupby('customer_id').order_id.nunique()
repeat_rate = (customer_orders >= 2).mean()
print(f'Repeat customer rate: {repeat_rate:.1%}')
```
